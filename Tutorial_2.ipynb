```python
import numpy as np
```

1. Create a numpy array of size 2 x 2

```python
arr = np.array([[2,2],[3,4]])
print("This is my 2 x 2 array:")
print(arr)
```

2. Create a random numpy array of size 3 x 4

```python
arr_rand = np.random.rand(3, 4)
print(arr_rand)
```

3. Create a numpy array of ones and a numpy array of zeroes

```python
arr_ones = np.ones((6,6))
print("This is my array of ones:")
print(arr_ones)

arr_zeros = np.zeros((7,7))
print("This is my array of zeros:")
print(arr_zeros)
```

4. Create a random numpy array of size 6x3 and transpose it

```python
arr_rand = np.random.rand(6, 3)
print("Original 6x3 array:")
print(arr_rand)

arr_rand_trans = arr_rand.T
print("Transposed 3x6 array:")
print(arr_rand_trans)
```

5. Create two random numpy array of size n x p and p x m and then compute the dot product

```python
arr_1 = np.random.rand(3,4)
arr_2 = np.random.rand(4,5)
print("This is the first array: ")
print(arr_1)
print("This is the second array: ")
print(arr_2)
print("This is the dot product of them: ")
arr_3 = np.dot(arr_1, arr_2)
print(arr_3)
```
